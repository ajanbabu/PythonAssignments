```python
import pandas as pd
```

```python
dataset= pd.read_csv("50_startups.csv")
```

```python
dataset
```

```python
dataset = pd.get_dummies(dataset).astype(int)
```

```python
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administartion', 'Mrketing Spend',
       'State_New York']]
```

```python
independent
```

```python
dependent = dataset[['Profit']]
```

```python
dependent
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(independent, dependent,test_size=0.30,random_state=0)
```

```python
X_train
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor=RandomForestRegressor(n_estimators=1000,criterion='squared_error', random_state = 1.0)
regressor.fit(X_train,Y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score=r2_score(Y_test,y_pred)
r2_score
```

```python
r2_score
```

```python
import pickle 
filename="finalized_model_random_forest.sav"
```

```python
pickle.dump(regressor,open(filename,'wb'))
```
