```python
#Print Odd numbers in the list
```

```python
numbers = [20,10,16,19,25,1,276,188]
for odd in numbers:
    if(odd%2 != 0):
        print(odd ,"is odd")
```

```python
# Print Even numbers in the list
```

```python
numbers = [20,10,16,19,25,1,276,188]
for even in numbers:
    if(even%2 == 0):
        print(even ,"is even")
```
