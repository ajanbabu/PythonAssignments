```python
#Create a function that checks whether the given number is odd or even
```

```python
class oddEven():
    def oddOrEven():
        number = int(input("Enter a number:"))
        if(number % 2 == 0):
            print(number,"is Even number")
        else:
            print(number,"is Odd number")
```

```python
oddEven.oddOrEven()
```

```python
# Create a function that tells elegibility of marriage for male and female according to their age limit like 21 for male and 18 for female
```

```python
class elegibilityForMarriage():
    def elegibility():
        gender = input("Your Gender:")
        age = int(input("Your Age:"))
        if(gender == 'male' and age>=21):
            print("Elegibile")
        elif(gender=='female' and age>=18):
            print("Elegibile")
        else:
            print("Not Elegibile")
        
```

```python
elegibilityForMarriage.elegibility()
```

```python
#Calculate the percentage of your 10th mark
```

```python
class findPercentage():
    def percentage():
        marks=[]

        for i in range(1,6):
            mark= int(input(f"Enter Subject {i} marks: "))
            marks.append(mark)
            
        Total = sum(marks)
        percentage = (Total/500)*100
        print("Total:",Total)
        print("percentage:",percentage)
        
        
```

```python
findPercentage.percentage()
```

```python
#Print area and perimeter of triangle using class and function
```

```python
class triangle():
    def areaOfTriangle():
        height = 32
        breadth = 34
        areaFormula = (height*breadth)/2
        print("Area of Triangle:",areaFormula)

    def perimeter():
        height1=2
        height2=4
        breadth=4
        perimeterOfTriangle = height1+height2+breadth
        print("perimeter of Triangle:",perimeterOfTriangle)
        
        
```

```python
triangle.areaOfTriangle()
triangle.perimeter()
```
